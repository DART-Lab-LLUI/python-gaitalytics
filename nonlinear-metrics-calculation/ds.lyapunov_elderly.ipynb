## Batch 1

```python
import pandas as pd
import numpy as np
import xarray as xr
import matplotlib.pyplot as plt
from scipy.stats import gaussian_kde
from pathlib import Path

PATIENTS = [
    ("E03", "Before"),
    ("E04", "Before"),
    ("E05", "Before"),
    ("E08", "Before"),
    ("E09", "Before"),
    ("E11", "Before"),
    ("E12", "Before"),
    ("E14", "Before"),
    ("E15", "Before"),
    ("E16", "Before"),
    ("E17", "Before"),
    # ("E30", "Before"),
    # ("E31", "Before"),
    # ("E32", "Before"),
    # ("E33", "Before"),
    # ("E34", "Before"),
    # ("E35", "Before"),
    # ("E36", "Before"),
    # ("E37", "Before"),
    # ("E38", "Before")

]
    
# Global parameters 
fs = 60.0  
max_lag = 30
bins = 64  
max_dim = 20  
rtol = 0.10  
fnn_threshold = 0.80 
max_time = 5.0 

# Initialize results dataframe
results_df = pd.DataFrame(columns=[
    'subject', 'condition', 'axis', 'average stride duration',
    'measuring time (s)', 'lag', 'embedding_dimension', 
    'fnn_threshold', 'lyapunov_exponent'
])

# Process each patient
for patient_id, condition in PATIENTS:
    print(f"\n{'*'*60}")
    print(f"Processing Patient {patient_id} - {condition}")
    print(f"{'*'*60}")
    patient_num = patient_id[1:] 
   
    try:
        # Construct paths
        file_path =   'Elderly' / patient_num / 'markers.nc'
        events_path =   'Elderly' /patient_num/ 'events.nc'
        
        # Load dataset
        ds = xr.open_dataset(file_path)
        da = ds['markers'] if 'markers' in ds else ds
        com = da.sel(channel='COM')
        time = com.coords['time'].values
        print('Time coordinates:', time[0], 'to', time[-1])
        
        # Get foot strike times
        fstrike_time = get_foot_strike(events_path, foot='Right', start_time=30.0, end_time=120.0)
        start_time = fstrike_time['first_strike']
        end_time = fstrike_time['last_strike']
        
        
        # Preprocess 
        com_preprocessed = {}
        for axis in ["x", "z"]:
            print(f"Preprocessing axis {axis}...")
            ts_raw = com.sel(axis=axis).values.ravel()
            # Downsampling
            ts_processed = preprocess_signal(ts_raw, fs=100, poly_order=1, hp_cutoff=0.1, lp_cutoff=30.0, target_fs=60)
            com_preprocessed[axis] = ts_processed
        
        
       
        time_array_100hz = com.coords['time'].values
        start_idx_100hz = np.argmin(np.abs(time_array_100hz - start_time))
        end_idx_100hz = np.argmin(np.abs(time_array_100hz - end_time))
    
        
        start_idx = int(start_idx_100hz  * 60.0 / 100.0)
        end_idx = int(end_idx_100hz  * 60.0 / 100.0)
        
        print(f'Selected time range: {start_time} to {end_time}')
        print(f'Selected indices at 50Hz: {start_idx} to {end_idx}')
        
        # Slice the preprocessed data
        com_sliced = {}
        for axis in ["x", "z"]:
            com_sliced[axis] = com_preprocessed[axis][start_idx:end_idx]
        
        print(f'Sliced COM shape: {com_sliced["x"].shape}')

        # Get average stride duration 
        average_stride = get_average_stride_duration(events_path, foot='Right', start_time=30.0, end_time=120.0)
        print('Average stride duration:', average_stride['mean_duration'], '±', average_stride['std_duration'], 's')
        
       
        max_lag_patient = max_lag
        max_dim_patient = max_dim
        max_time_patient = 5.0 
        
        # Process each axis
        for axis in ['x', 'z']:
            print(f'\n--- Axis: {axis} ---')
            ts = com_sliced[axis]
            
            # Histogram
            plt.figure(figsize=(6, 4))
            plt.hist(ts, bins=50, density=True, edgecolor='k', alpha=0.6)
            plt.xlabel(f'COM {axis} value')
            plt.ylabel('Probability density')
            plt.title(f'Histogram of COM_{axis} (preprocessed, 50Hz)')
            plt.grid(True)
            plt.show()
            
            # KDE overlay - visual check
            kde = gaussian_kde(ts)
            xs = np.linspace(ts.min(), ts.max(), 500)
            plt.figure(figsize=(6, 4))
            plt.hist(ts, bins=50, density=True, alpha=0.4, edgecolor='k', label='Histogram')
            plt.plot(xs, kde(xs), 'r-', lw=2, label='KDE')
            plt.xlabel(f'COM {axis} value')
            plt.ylabel('Probability density')
            plt.title(f'Histogram + KDE of COM {axis} (preprocessed, 50Hz)')
            plt.legend()
            plt.grid(True)
            plt.show()
            
            # Find optimal lag via AMI
            lag = optimal_lag(ts, fs=fs, max_lag=max_lag_patient, bins=bins)
            print(f'Optimal lag (τ) for axis {axis}: {lag} samples ({lag/fs:.3f} s)')
            
            # Compute FNN% up to max_dim
            dims, fnn_perc = compute_fnn_percentage(ts, lag, max_dim=max_dim_patient, rtol=rtol)
            
            # Plot FNN% vs. embedding dimension
            plt.figure()
            plt.plot(dims, fnn_perc, '-o', label=f'Axis {axis}')
            plt.axhline(fnn_threshold, linestyle='--', color='k', label=f'{int(fnn_threshold*100)}% Threshold')
            plt.xlabel('Embedding dimension (m)')
            plt.ylabel('False Nearest Neighbors (%)')
            plt.title(f'FNN% vs. m for axis {axis} (lag={lag})')
            plt.grid(True)
            plt.legend()
            plt.show()
            
            # Find optimal embedding dimension
            dim = optimal_embedding_dimension(ts, lag, max_dim=max_dim_patient, rtol=rtol)
            print(f'Optimal embedding dimension (m) for axis {axis}: {dim}')
            
            # Compute Lyapunov exponent 
            lyap = compute_lyapunov_exponent(
                ts,
                emb_dim=dim,
                lag=lag,
                fs=fs, 
                max_time=max_time_patient,
                fit_start=0.0,
                fit_end=average_stride['mean_duration'],
                plot_residuals=True
            )
            print(f'Lyapunov exponent (λ*) for axis {axis}: {lyap:.4f} 1/s')
            
            # Store results
            results_df.loc[len(results_df)] = {
                'subject': patient_id,
                'condition': condition,
                'axis': axis,
                'average stride duration': average_stride['mean_duration'],
                'measuring time (s)': end_time - start_time,
                'lag': lag,
                'embedding_dimension': dim,
                'fnn_threshold': fnn_threshold,
                'lyapunov_exponent': lyap
            }
            
            print(results_df)
        
        # Close dataset
        ds.close()
       
        # Print summary for patient
        print(f"\n=== SUMMARY - Patient {patient_id} ===")
        print(f"Condition: {condition}")
        print(f'Analysis duration: {end_time - start_time:.2f} seconds')
        print(f'Average stride duration: {average_stride["mean_duration"]:.3f} ± {average_stride["std_duration"]:.3f} s')
        
        # Print Lyapunov results
        patient_results = results_df[results_df['subject'] == patient_id].tail(2)
        for _, row in patient_results.iterrows():
            print(f"\n{row['axis'].upper()}-axis:")
            print(f"  Lag (τ): {row['lag']} samples ({row['lag']/fs:.3f} s)")
            print(f"  Embedding dimension (m): {row['embedding_dimension']}")
            print(f"  Lyapunov exponent (λ*): {row['lyapunov_exponent']:.4f} 1/s")
    
    except Exception as e:
        print(f"ERROR processing patient {patient_id}: {str(e)}")
        continue

# Display final results
print("\n" + "="*60)
print("FINAL RESULTS - ALL PATIENTS")
print("="*60)
print(results_df)

```

```python
import pandas as pd
import numpy as np
import xarray as xr
import matplotlib.pyplot as plt
from scipy.stats import gaussian_kde
from pathlib import Path

# Define patient configurations
PATIENTS = [
    ("E03", "After"),
    ("E04", "After"),
    ("E05", "After"),
    ("E08", "After"),
    ("E09", "After"),
    ("E11", "After"),
    ("E12", "After"),
    ("E14", "After"),
    ("E15", "After"),
    ("E16", "After"),
    ("E17", "After"),
    # ("E30", "After"),
    # ("E31", "After"),
    # ("E32", "After"),
    # ("E33", "After"),
    # ("E34", "After"),
    # ("E35", "After"),
    # ("E36", "After"),
    # ("E37", "After"),
    # ("E38", "After")

]

# Global parameters - UPDATED FOR 60 Hz
fs = 60.0  # Changed from 100.0 to 60.0
max_lag = 30  # Changed from 20 to 30 (0.5s at 60Hz)
bins = 64  # Keep the same
max_dim = 15  # Keep the same
rtol = 0.10  # Keep the same
fnn_threshold = 0.80  # Keep the same
max_time = 5.0  # Keep the same (this is in seconds, not samples)

# Initialize results dataframe - UNCOMMENTED AND FIXED
results_df = pd.DataFrame(columns=[
    'subject', 'condition', 'axis', 'average stride duration',
    'measuring time (s)', 'lag', 'embedding_dimension', 
    'fnn_threshold', 'lyapunov_exponent'
])

# Process each patient
for patient_id, condition in PATIENTS:
    print(f"\n{'*'*60}")
    print(f"Processing Patient {patient_id} - {condition}")
    print(f"{'*'*60}")
    
    try:
    
        if patient_id[1:] == '03':
            skip_start = 139.5
        elif patient_id[1:] == '04':
            skip_start =  120.5
        elif patient_id[1:] == '05':
            skip_start =  126.0
        elif patient_id[1:] == '08':
            skip_start =  151.5
        elif patient_id[1:] == '09':
            skip_start = 124.0
        elif patient_id[1:] == '11':
            skip_start =  139.0
        elif patient_id[1:] == '12':
            skip_start =  126.5
        elif patient_id[1:] == '14':
            skip_start = 130.5
        elif patient_id[1:] == '15':
            skip_start =  135.5
        elif patient_id[1:] == '16':
            skip_start =  123.2
        elif patient_id[1:] == '17':
            skip_start =  126.2
        elif patient_id[1:]=='30':
            skip_start = 125.0
        elif patient_id[1:] == '31':
            skip_start =  122.0
        elif patient_id[1:] == '32':
            skip_start =  125.5
        elif patient_id[1:] == '33':
            skip_start =  139.0
        elif patient_id[1:] == '34':
            skip_start =  124.0
        elif patient_id[1:] == '35':    
            skip_start =  128.0
        elif patient_id[1:] == '36':    
            skip_start =  125.5
        elif patient_id[1:] == '37':    
            skip_start =  126.5
        elif patient_id[1:] == '38':
            skip_start =  125.5
        else :
            skip_start = 160.0
            
        print(f"Using skip_start time: {skip_start}s")
        patient_num = patient_id[1:] 
        # Construct paths
        file_path = cwd.parent / '..' / 'data' / 'PBT' / 'Elderly' / patient_num / 'markers.nc'
        events_path = cwd.parent / '..' / 'data' / 'PBT' /  'Elderly' / patient_num /  'events.nc'
        
        # Load dataset
        ds = xr.open_dataset(file_path)
        da = ds['markers'] if 'markers' in ds else ds
        com = da.sel(channel='COM')
        time_array = com.coords['time'].values
        print('Time coordinates:', time_array[0], 'to', time_array[-1])
        
        # Calculate end_time for foot strike search
        end_time_search = time_array[-1] - 5.0
        
        # Get foot strike times using skip_start
        fstrike_time = get_foot_strike(events_path, foot='Right', start_time=skip_start, end_time=end_time_search)
        start_time = fstrike_time['first_strike']
        end_time = fstrike_time['last_strike']
        
        print(f'Foot strikes found from {start_time} to {end_time}')
        
        # ------------- PREPROCESS BEFORE SLICING -------------
        # Preprocess each axis of the full COM data
        com_preprocessed = {}
        for axis in ["x", "z"]:
            print(f"Preprocessing axis {axis}...")
            ts_raw = com.sel(axis=axis).values.ravel()
            # Downsampling to 60 Hz happens here
            ts_processed = preprocess_signal(ts_raw, fs=100, poly_order=1, hp_cutoff=0.1, lp_cutoff=30.0, target_fs=60)
            com_preprocessed[axis] = ts_processed
        
        # IMPORTANT: After downsampling, time indices need to be adjusted
        # Original time array at 100 Hz
        time_array_100hz = com.coords['time'].values
        start_idx_100hz = np.argmin(np.abs(time_array_100hz - start_time))
        end_idx_100hz = np.argmin(np.abs(time_array_100hz - end_time))
        
        # Convert indices to 60 Hz
        start_idx = int(start_idx_100hz * 60.0 / 100.0)
        end_idx = int(end_idx_100hz * 60.0 / 100.0)
        
        print(f'Selected time range: {start_time} to {end_time}')
        print(f'Selected indices at 60Hz: {start_idx} to {end_idx}')
        
        # Slice the preprocessed data
        com_sliced = {}
        for axis in ["x", "z"]:
            com_sliced[axis] = com_preprocessed[axis][start_idx:end_idx]
        
        print(f'Sliced COM shape: {com_sliced["x"].shape}')

        # Get average stride duration (this is in seconds, so no change needed)
        average_stride = get_average_stride_duration(events_path, foot='Right', start_time=skip_start, end_time=end_time_search-5.0)
        print('Average stride duration:', average_stride['mean_duration'], '±', average_stride['std_duration'], 's')
        
        # Set patient-specific parameters - UPDATED FOR 60 Hz
        max_lag_patient = max_lag  # Use global max_lag = 30
        max_dim_patient = max_dim  # Use global max_dim = 15
        max_time_patient = 5.0  # Can be different from global if needed
        
        # Process each axis
        for axis in [ 'x','z']:
            print(f'\n--- Axis: {axis} ---')
            ts = com_sliced[axis]
            
            # Histogram
            plt.figure(figsize=(6, 4))
            plt.hist(ts, bins=50, density=True, edgecolor='k', alpha=0.6)
            plt.xlabel(f'COM {axis} value')
            plt.ylabel('Probability density')
            plt.title(f'Histogram of COM_{axis} (preprocessed, 60Hz)')
            plt.grid(True)
            plt.show()
            
            # KDE overlay
            kde = gaussian_kde(ts)
            xs = np.linspace(ts.min(), ts.max(), 500)
            plt.figure(figsize=(6, 4))
            plt.hist(ts, bins=50, density=True, alpha=0.4, edgecolor='k', label='Histogram')
            plt.plot(xs, kde(xs), 'r-', lw=2, label='KDE')
            plt.xlabel(f'COM {axis} value')
            plt.ylabel('Probability density')
            plt.title(f'Histogram + KDE of COM {axis} (preprocessed, 60Hz)')
            plt.legend()
            plt.grid(True)
            plt.show()
            
            # Find optimal lag via AMI - using global parameters
            lag = optimal_lag(ts, fs=fs, max_lag=max_lag_patient, bins=bins)
            print(f'Optimal lag (τ) for axis {axis}: {lag} samples ({lag/fs:.3f} s)')
            
            # Compute FNN% up to max_dim
            dims, fnn_perc = compute_fnn_percentage(ts, lag, max_dim=max_dim_patient, rtol=rtol)
            
            # Plot FNN% vs. embedding dimension
            plt.figure()
            plt.plot(dims, fnn_perc, '-o', label=f'Axis {axis}')
            plt.axhline(fnn_threshold, linestyle='--', color='k', label=f'{int(fnn_threshold*100)}% Threshold')
            plt.xlabel('Embedding dimension (m)')
            plt.ylabel('False Nearest Neighbors (%)')
            plt.title(f'FNN% vs. m for axis {axis} (lag={lag})')
            plt.grid(True)
            plt.legend()
            plt.show()
            
            # Find optimal embedding dimension
            dim = optimal_embedding_dimension(ts, lag, max_dim=max_dim_patient, rtol=rtol)
            print(f'Optimal embedding dimension (m) for axis {axis}: {dim}')
            
            # Compute Lyapunov exponent - fs is now 60
            lyap = compute_lyapunov_exponent(
                ts,
                emb_dim=dim,
                lag=lag,
                fs=fs,  # Now 60 Hz
                max_time=max_time_patient,
                fit_start=0.0,
                fit_end=average_stride['mean_duration'],
                plot_residuals=True
            )
            print(f'Lyapunov exponent (λ*) for axis {axis}: {lyap:.4f} 1/s')
            
            # Store results
            results_df.loc[len(results_df)] = {
                'subject': patient_id,
                'condition': condition,
                'axis': axis,
                'average stride duration': average_stride['mean_duration'],
                'measuring time (s)': end_time - start_time,
                'lag': lag,
                'embedding_dimension': dim,
                'fnn_threshold': fnn_threshold,
                'lyapunov_exponent': lyap
            }
            
            print(results_df)
        
        # Close dataset
        ds.close()
       
        # Print summary for this patient
        print(f"\n=== SUMMARY - Patient {patient_id} ===")
        print(f"Condition: {condition}")
        print(f'Analysis duration: {end_time - start_time:.2f} seconds')
        print(f'Average stride duration: {average_stride["mean_duration"]:.3f} ± {average_stride["std_duration"]:.3f} s')
        
        # Print Lyapunov results
        patient_results = results_df[results_df['subject'] == patient_id].tail(2)
        for _, row in patient_results.iterrows():
            print(f"\n{row['axis'].upper()}-axis:")
            print(f"  Lag (τ): {row['lag']} samples ({row['lag']/fs:.3f} s)")
            print(f"  Embedding dimension (m): {row['embedding_dimension']}")
            print(f"  Lyapunov exponent (λ*): {row['lyapunov_exponent']:.4f} 1/s")
    
    except Exception as e:
        print(f"ERROR processing patient {patient_id}: {str(e)}")
        import traceback
        traceback.print_exc()
        continue

# Display final results
print("\n" + "="*60)
print("FINAL RESULTS - ALL PATIENTS")
print("="*60)
print(results_df)
```

```python

PATIENTS = [
    ("E03", "After"),
    ("E04", "After"),
    ("E05", "After"),
    ("E08", "After"),
    ("E09", "After"),
    ("E11", "After"),
    ("E12", "After"),
    ("E14", "After"),
    ("E15", "After"),
    ("E16", "After"),
    ("E17", "After"),
    # ("E30", "After"),
    # ("E31", "After"),
    # ("E32", "After"),
    # ("E33", "After"),
    # ("E34", "After"),
    # ("E35", "After"),
    # ("E36", "After"),
    # ("E37", "After"),
    # ("E38", "After")

]

# Global parameters
fs = 60.0  
max_lag = 30  
bins = 64  
max_dim = 15 
rtol = 0.10  
fnn_threshold = 0.80  
max_time = 5.0  

# Initialize results dataframe 
results_df = pd.DataFrame(columns=[
    'subject', 'condition', 'axis', 'average stride duration',
    'measuring time (s)', 'lag', 'embedding_dimension', 
    'fnn_threshold', 'lyapunov_exponent'
])

# Process each patient
for patient_id, condition in PATIENTS:
    print(f"\n{'*'*60}")
    print(f"Processing Patient {patient_id} - {condition}")
    print(f"{'*'*60}")
    
    try:
        patient_num = patient_id[1:]
        if patient_id[1:] == '03':
                skip_start = 121.5
        elif patient_id[1:] == '04':
            skip_start =  121.8
        elif patient_id[1:] == '05':
            skip_start =  121.8
        elif patient_id[1:] == '08':
            skip_start =  122.6
        elif patient_id[1:] == '09':
            skip_start = 123.0
        elif patient_id[1:] == '11':
            skip_start =  121.0
        elif patient_id[1:] == '12':
            skip_start =  125.0
        elif patient_id[1:] == '14':
            skip_start = 127.0
        elif patient_id[1:] == '15':
            skip_start =  125.0
        elif patient_id[1:] == '16':
            skip_start =  125.0
        elif patient_id[1:] == '17':
            skip_start =  125.0
        elif patient_id[1:]=='30':
            skip_start = 125.0
        elif patient_id[1:] == '31':
            skip_start =  123.0
        elif patient_id[1:] == '32':
            skip_start =  124.0
        elif patient_id[1:] == '33':
            skip_start =  123.5
        elif patient_id[1:] == '34':
            skip_start =  124.0
        elif patient_id[1:] == '35':    
            skip_start =  124.0
        elif patient_id[1:] == '36':    
            skip_start =  124.5
        elif patient_id[1:] == '37':    
            skip_start =  124.5
        elif patient_id[1:] == '38':
            skip_start =  123.0
        else :
            skip_start = 160.0
        # if patient_id[1:] == '03':
        #     skip_start = 139.5
        # elif patient_id[1:] == '04':
        #     skip_start =  122.5
        # elif patient_id[1:] == '05':
        #     skip_start =  123.8
        # elif patient_id[1:] == '08':
        #     skip_start =  154.0
        # elif patient_id[1:] == '09':
        #     skip_start = 124.0
        # elif patient_id[1:] == '11':
        #     skip_start =  121.0
        # elif patient_id[1:] == '12':
        #     skip_start =  127.0
        # elif patient_id[1:] == '14':
        #     skip_start = 130.5
        # elif patient_id[1:] == '15':
        #     skip_start =  135.5
        # elif patient_id[1:] == '16':
        #     skip_start =  123.0
        # elif patient_id[1:] == '17':
        #     skip_start =  140.0
        # elif patient_id[1:]=='30':
        #     skip_start = 125.0
        # elif patient_id[1:] == '31':
        #     skip_start =  123.0
        # elif patient_id[1:] == '32':
        #     skip_start =  125.5
        # elif patient_id[1:] == '33':
        #     skip_start =  127.0
        # elif patient_id[1:] == '34':
        #     skip_start =  124.0
        # elif patient_id[1:] == '35':    
        #     skip_start =  126.5
        # elif patient_id[1:] == '36':    
        #     skip_start =  124.5
        # elif patient_id[1:] == '37':    
        #     skip_start =  126.5
        # elif patient_id[1:] == '38':
        #     skip_start =  125.5
        # else :
        #     skip_start = 160.0
            
        print(f"Using skip_start time: {skip_start}s")
        patient_num = patient_id[1:] 
        # Construct paths
        file_path = 'Elderly' / patient_num / 'markers.nc'
        events_path =   'Elderly' / patient_num /  'events.nc'
        
        # Load dataset
        ds = xr.open_dataset(file_path)
        da = ds['markers'] if 'markers' in ds else ds
        com = da.sel(channel='COM')
        time_array = com.coords['time'].values
        print('Time coordinates:', time_array[0], 'to', time_array[-1])
        
        # Calculate end_time for foot strike search
        end_time_search = time_array[-1] - 5.0
        
        # Get foot strike times using skip_start
        fstrike_time = get_foot_strike(events_path, foot='Right', start_time=skip_start, end_time=end_time_search)
        start_time = fstrike_time['first_strike']
        end_time = fstrike_time['last_strike']
        
        print(f'Foot strikes found from {start_time} to {end_time}')
        
     
        com_preprocessed = {}
        for axis in ["x", "z"]:
            print(f"Preprocessing axis {axis}...")
            ts_raw = com.sel(axis=axis).values.ravel()
            # Downsampling to 60 Hz happens here
            ts_processed = preprocess_signal(ts_raw, fs=100, poly_order=1, hp_cutoff=0.1, lp_cutoff=30.0, target_fs=60)
            com_preprocessed[axis] = ts_processed
        
        time_array_100hz = com.coords['time'].values
        start_idx_100hz = np.argmin(np.abs(time_array_100hz - start_time))
        end_idx_100hz = np.argmin(np.abs(time_array_100hz - end_time))
        
        
        start_idx = int(start_idx_100hz * 60.0 / 100.0)
        end_idx = int(end_idx_100hz * 60.0 / 100.0)
        
        print(f'Selected time range: {start_time} to {end_time}')
        print(f'Selected indices at 60Hz: {start_idx} to {end_idx}')
        
        # Slice the preprocessed data
        com_sliced = {}
        for axis in ["x", "z"]:
            com_sliced[axis] = com_preprocessed[axis][start_idx:end_idx]
        
        print(f'Sliced COM shape: {com_sliced["x"].shape}')

        # Get average stride duration 
        average_stride = get_average_stride_duration(events_path, foot='Right', start_time=skip_start, end_time=end_time_search-5.0)
        print('Average stride duration:', average_stride['mean_duration'], '±', average_stride['std_duration'], 's')
        
     
        max_lag_patient = max_lag  
        max_dim_patient = max_dim 
        max_time_patient = 5.0 
        
        # Process each axis
        for axis in [ 'x','z']:
            print(f'\n--- Axis: {axis} ---')
            ts = com_sliced[axis]
            
            # Histogram
            plt.figure(figsize=(6, 4))
            plt.hist(ts, bins=50, density=True, edgecolor='k', alpha=0.6)
            plt.xlabel(f'COM {axis} value')
            plt.ylabel('Probability density')
            plt.title(f'Histogram of COM_{axis} (preprocessed, 60Hz)')
            plt.grid(True)
            plt.show()
            
            # KDE overlay - visual check
            kde = gaussian_kde(ts)
            xs = np.linspace(ts.min(), ts.max(), 500)
            plt.figure(figsize=(6, 4))
            plt.hist(ts, bins=50, density=True, alpha=0.4, edgecolor='k', label='Histogram')
            plt.plot(xs, kde(xs), 'r-', lw=2, label='KDE')
            plt.xlabel(f'COM {axis} value')
            plt.ylabel('Probability density')
            plt.title(f'Histogram + KDE of COM {axis} (preprocessed, 60Hz)')
            plt.legend()
            plt.grid(True)
            plt.show()
            
            # Find optimal lag via AMI
            lag = optimal_lag(ts, fs=fs, max_lag=max_lag_patient, bins=bins)
            print(f'Optimal lag (τ) for axis {axis}: {lag} samples ({lag/fs:.3f} s)')
            
            # Compute FNN% up to max_dim
            dims, fnn_perc = compute_fnn_percentage(ts, lag, max_dim=max_dim_patient, rtol=rtol)
            
            # Plot FNN% vs. embedding dimension
            plt.figure()
            plt.plot(dims, fnn_perc, '-o', label=f'Axis {axis}')
            plt.axhline(fnn_threshold, linestyle='--', color='k', label=f'{int(fnn_threshold*100)}% Threshold')
            plt.xlabel('Embedding dimension (m)')
            plt.ylabel('False Nearest Neighbors (%)')
            plt.title(f'FNN% vs. m for axis {axis} (lag={lag})')
            plt.grid(True)
            plt.legend()
            plt.show()
            
            # Find optimal embedding dimension
            dim = optimal_embedding_dimension(ts, lag, max_dim=max_dim_patient, rtol=rtol)
            print(f'Optimal embedding dimension (m) for axis {axis}: {dim}')
            
            # Compute Lyapunov exponent 
            lyap = compute_lyapunov_exponent(
                ts,
                emb_dim=dim,
                lag=lag,
                fs=fs, 
                max_time=max_time_patient,
                fit_start=0.0,
                fit_end=average_stride['mean_duration'],
                plot_residuals=True
            )
            print(f'Lyapunov exponent (λ*) for axis {axis}: {lyap:.4f} 1/s')
            
            # Store results
            results_df.loc[len(results_df)] = {
                'subject': patient_id,
                'condition': condition,
                'axis': axis,
                'average stride duration': average_stride['mean_duration'],
                'measuring time (s)': end_time - start_time,
                'lag': lag,
                'embedding_dimension': dim,
                'fnn_threshold': fnn_threshold,
                'lyapunov_exponent': lyap
            }
            
            print(results_df)
        
        # Close dataset
        ds.close()
       
        # Print summary for this patient
        print(f"\n=== SUMMARY - Patient {patient_id} ===")
        print(f"Condition: {condition}")
        print(f'Analysis duration: {end_time - start_time:.2f} seconds')
        print(f'Average stride duration: {average_stride["mean_duration"]:.3f} ± {average_stride["std_duration"]:.3f} s')
        
        # Print Lyapunov results
        patient_results = results_df[results_df['subject'] == patient_id].tail(2)
        for _, row in patient_results.iterrows():
            print(f"\n{row['axis'].upper()}-axis:")
            print(f"  Lag (τ): {row['lag']} samples ({row['lag']/fs:.3f} s)")
            print(f"  Embedding dimension (m): {row['embedding_dimension']}")
            print(f"  Lyapunov exponent (λ*): {row['lyapunov_exponent']:.4f} 1/s")
    
    except Exception as e:
        print(f"ERROR processing patient {patient_id}: {str(e)}")
        import traceback
        traceback.print_exc()
        continue

# Display final results
print("\n" + "="*60)
print("FINAL RESULTS - ALL PATIENTS")
print("="*60)
print(results_df)
```
